# Chủ Đề 9: Session-Based Recommendation System
## Hệ Thống Khuyến Nghị Dựa Trên Hành Vi Duyệt Web

---

### 1. Bài toán thực tế

Hãy tưởng tượng bạn là khách hàng **ẩn danh** truy cập trang web bán sách:
- Bạn **không đăng nhập**, không có tài khoản
- Trong 15 phút, bạn xem lần lượt: `Đắc Nhân Tâm` → `Nghĩ Giàu Làm Giàu` → `Bí Quyết Tư Duy Triệu Phú`
- Hệ thống CF truyền thống **bó tay** vì không biết bạn là ai

**Session-based RS** giải quyết chính xác tình huống này:
> Dự đoán sản phẩm tiếp theo người dùng sẽ muốn xem, chỉ dựa vào chuỗi hành động trong phiên hiện tại.

### 2. Định nghĩa hình thức

- **Session (phiên):** S = [i₁, i₂, i₃, ..., iₜ] – chuỗi sản phẩm đã tương tác theo thứ tự thời gian
- **Mục tiêu:** Cho S = [i₁, i₂, ..., iₜ], dự đoán iₜ₊₁
- Đây là bài toán **Next-Item Prediction**

```
INPUT:  [item_1, item_2, item_3]   ← chuỗi đã click trong phiên
OUTPUT: item_4                      ← dự đoán item tiếp theo
```

### 3. So sánh với Collaborative Filtering

| Tiêu chí | Collaborative Filtering | Session-based RS |
|---|---|---|
| Cần ID người dùng? | Bắt buộc | **Không cần** |
| Dùng lịch sử dài hạn? | Có (toàn bộ) | **Không** (chỉ phiên hiện tại) |
| Cold-start user mới? | Không giải quyết được | **Tự nhiên xử lý được** |
| Ứng dụng | Netflix, Spotify | Amazon, Shopee (khách ẩn danh) |

### 4. Các thuật toán sẽ cài đặt

| # | Thuật toán | Ý tưởng | Độ khó |
|---|---|---|---|
| 1 | **Popularity Baseline** | Gợi ý item phổ biến nhất | Rất dễ |
| 2 | **SKNN** (Session KNN) | Tìm phiên cũ giống nhất → gợi ý item của phiên đó | Dễ |
| 3 | **GRU4Rec** | Dùng mạng GRU đọc chuỗi click → dự đoán item tiếp | Trung bình |

### 5. Dataset sử dụng

**Yoochoose** (RecSys Challenge 2015)
- 9 triệu phiên, 53 triệu click
- Dữ liệu thương mại điện tử thực tế
- Sử dụng bản 1/64 (~140K phiên) để chạy trên laptop
